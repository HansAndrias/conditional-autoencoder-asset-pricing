```python
import pandas as pd
from pathlib import Path

# Mount Google Drive
from google.colab import drive
drive.mount('/content/drive')

# Define the base path for Google Drive
base_path = Path('/content/drive/My Drive/Tidy Finance ML dataset/Robustness_check_predictive')

# Define the results path relative to the base path
results_path = base_path / 'results' / 'asset_pricing'

# Load the fold scores from the HDF5 file
scores_file_path = str(results_path / 'scores.h5')

# Read the HDF5 file containing the scores
with pd.HDFStore(scores_file_path, mode='r') as store:
    keys = store.keys()  # Get all the keys in the HDF5 file
    fold_scores = {key: store[key] for key in keys}  # Load all dataframes

# Combine all fold scores into a single DataFrame
all_folds_df = pd.concat(fold_scores.values())

# Calculate the average across all folds for each num_hidden_layers
average_scores = all_folds_df.groupby('num_hidden_layers').mean().reset_index()

# Display the average scores
print("Average scores across all folds for each num_hidden_layers:")
print(average_scores)

# Convert the average scores DataFrame to a LaTeX table
latex_table = average_scores.to_latex(index=False, float_format="%.6f")

# Display the LaTeX table
print("LaTeX Table:")
print(latex_table)
```

#**PLS MODEL**

```python
import pandas as pd
from pathlib import Path

# Step 1: Mount Google Drive
drive.mount('/content/drive')

# Define the base path for Google Drive
base_path = Path('/content/drive/My Drive/Tidy Finance ML dataset/Robustness_check_predictive_PLS')

# Define the results path relative to the base path
results_path = base_path / 'results' / 'asset_pricing'

# Load the R² scores for the main model
r2_scores_file = results_path / 'r2_scores.h5'
r2_xt_scores_file = results_path / 'r2_xt_scores.h5'

if r2_scores_file.exists() and r2_xt_scores_file.exists():
    # Load the R² scores for the main model
    r2_scores_df = pd.read_hdf(r2_scores_file, key='fold')
    avg_r2 = r2_scores_df['r2_score'].mean()

    # Load the R² scores for xt
    r2_xt_scores_df = pd.read_hdf(r2_xt_scores_file, key='fold')
    avg_r2_xt = r2_xt_scores_df['avg_r2_xt'].mean()

    # Create a DataFrame with the required structure
    final_df = pd.DataFrame({
        'Model': ['PLS'],
        'r2_score': [avg_r2],
        'avg_r2_xt': [avg_r2_xt]
    })

    # Convert the DataFrame to a LaTeX table
    latex_table = final_df.to_latex(index=False, float_format="%.6f")

    # Display the LaTeX table
    print("LaTeX Table:")
    print(latex_table)

else:
    if not r2_scores_file.exists():
        print(f"R² scores file not found: {r2_scores_file}")
    if not r2_xt_scores_file.exists():
        print(f"R² scores for xt file not found: {r2_xt_scores_file}")
```

```python
import pandas as pd
from pathlib import Path
import matplotlib.pyplot as plt
from google.colab import drive


# Step 1: Mount Google Drive (if running in Google Colab)
drive.mount('/content/drive')


# Define the base path for Google Drive
base_path = Path('/content/drive/My Drive/Tidy Finance ML dataset/Robustness_check_predictive_PLS')

# Define the results path relative to the base path
results_path = base_path / 'results' / 'asset_pricing'

# Check if the path exists, if not, create it
if not results_path.exists():
    results_path.mkdir(parents=True)

# Load the importance DataFrame from the HDF5 file
importances_df = pd.read_hdf(str(results_path / 'variable_importance_ipca.h5'), key='all_folds')

# Remove the prefix "characteristic_" from all characteristics
importances_df['variable'] = importances_df['variable'].str.replace('characteristic_', '', regex=False)

# Sort and select the top 20 variable importances
top_20_importances_df = importances_df.sort_values('importance_normalized', ascending=False).head(20)

# Reverse the DataFrame so that the highest importance is at the top in the plot
top_20_importances_df = top_20_importances_df.iloc[::-1]

# Plot the top 20 variable importances
top_20_importances_df.plot(kind='barh', x='variable', y='importance_normalized', legend=False)
plt.xlabel('Normalized Importance')
plt.ylabel('Variables')

# Save the plot to a file
plot_filename = 'top_20_importances_ipca.png'
plt.savefig(plot_filename, format='png')

# Show the plot (optional)
plt.show()

# Calculate average R² scores for PCA
r2_scores_file = results_path / 'r2_scores.h5'
r2_xt_scores_file = results_path / 'r2_xt_scores.h5'

if r2_scores_file.exists() and r2_xt_scores_file.exists():
    # Load the R² scores for the main model
    r2_scores_df = pd.read_hdf(r2_scores_file, key='fold')
    avg_r2 = r2_scores_df['r2_score'].mean()

    # Load the R² scores for xt
    r2_xt_scores_df = pd.read_hdf(r2_xt_scores_file, key='fold')
    avg_r2_xt = r2_xt_scores_df['avg_r2_xt'].mean()

    # Create a DataFrame with the required structure
    final_df = pd.DataFrame({
        'Model': ['PCA'],
        'r2_score': [avg_r2],
        'avg_r2_xt': [avg_r2_xt]
    })

    # Convert the DataFrame to a LaTeX table
    latex_table = final_df.to_latex(index=False, float_format="%.6f")

    # Display the LaTeX table
    print("LaTeX Table:")
    print(latex_table)

else:
    if not r2_scores_file.exists():
        print(f"R² scores file not found: {r2_scores_file}")
    if not r2_xt_scores_file.exists():
        print(f"R² scores for xt file not found: {r2_xt_scores_file}")
```

```python
import pandas as pd
from pathlib import Path
from google.colab import drive

# Step 1: Mount Google Drive
drive.mount('/content/drive')

# Define the base path for Google Drive
base_path = Path('/content/drive/My Drive/Tidy Finance ML dataset/Robustness_check_predictive_IPCA')

# Define the results path relative to the base path
results_path = base_path / 'results' / 'asset_pricing'

# Check if the path exists, if not, create it
if not results_path.exists():
    results_path.mkdir(parents=True)

# Load the R² scores for predictions
r2_scores_df = pd.read_hdf(str(results_path / 'r2_scores.h5'), key='fold')

# Load the average R² scores for xt
r2_xt_scores_df = pd.read_hdf(str(results_path / 'r2_xt_scores.h5'), key='fold')

# Calculate the average R² scores across all folds
avg_r2 = r2_scores_df['r2_score'].mean()
avg_r2_xt = r2_xt_scores_df['avg_r2_xt'].mean()

# Create a DataFrame with the required structure
final_df = pd.DataFrame({
    'Model': ['IPCA'],
    'r2_score': [avg_r2],
    'avg_r2_xt': [avg_r2_xt]
})

# Convert the DataFrame to a LaTeX table
latex_table = final_df.to_latex(index=False, float_format="%.6f")

# Display the LaTeX table
print("LaTeX Table:")
print(latex_table)
```
